```python
import pandas as pd

data = {
    "Case_ID": [101,102,103,104,105,106,107,108],

    "Drug": [
        "Paracetamol",
        "Ibuprofen",
        "Amoxicillin",
        "Metformin",
        "Aspirin",
        "Ciprofloxacin",
        "Azithromycin",
        "Diclofenac"
    ],

    "Reaction": [
        "Rash",
        "Nausea",
        "Headache",
        "Dizziness",
        "Vomiting",
        "Fever",
        "Skin Allergy",
        "Stomach Pain"
    ],

    "Severity": [
        "Mild",
        "Moderate",
        "Mild",
        "Severe",
        "Moderate",
        "Mild",
        "Severe",
        "Moderate"
    ],

    "Outcome": [
        "Recovered",
        "Recovering",
        "Recovered",
        "Not Recovered",
        "Recovered",
        "Recovered",
        "Recovering",
        "Recovered"
    ]
}

df = pd.DataFrame(data)

df
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
import matplotlib.pyplot as plt

df['Severity'].value_counts().plot(kind='bar')

plt.title("Adverse Event Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")

plt.show()
```

```python
df['Reaction'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title("Adverse Drug Reactions")

plt.ylabel("")

plt.show()
```
